# ETL
Todo o processo de limpeza e preparação do dataset.
- Limpeza e tokenização de perguntas
- Padronização e resize de imagens

#### Importando bibliotecas

```python
import pandas as pd
import numpy as np
from transformers import BertTokenizer, pipeline, BertConfig, BertModel
import torch
from datetime import datetime
```

```python
df = pd.read_json('datasets/dataset_pernambuco_25_turistas.json')

# df apenas de textos
df_textos = df[['id', 'pergunta', 'contexto']]

# df apenas ded imagens
df_imagens = df[['id', 'imagem', 'tags']]
```

## Limpeza e tokenização de textos

```python
# Carregando o tokenizador
tokenizer = BertTokenizer.from_pretrained('neuralmind/bert-base-portuguese-cased')

for i in range(len(df_textos['pergunta'])):
    pergunta = df_textos['pergunta'][i]

    tokens = tokenizer.tokenize(pergunta)
    encoding = tokenizer(
        pergunta,
        padding=True,
        do_lower_case=True,
        truncation=True,
        max_length=128,
        return_tensors='pt'
    )

    print("Input IDs:", encoding['input_ids'])
    print("Attention Mask:", encoding['attention_mask'])
```

## Testando BERT

```python
classificador_intencoes = pipeline(
    task='text-classification',
    model='neuralmind/bert-base-portuguese-cased',
    tokenizer='neuralmind/bert-base-portuguese-cased'
)

def chatbot_bert(pergunta):
    intencao = classificador_intencoes(pergunta)[0]
    
    if intencao['label'] == 'LABEL_0':
        return "Olá! Como posso ajudar você?"
    elif intencao['label'] == 'LABEL_1':
        return f"São {datetime.now().strftime('%H:%M')}"
    elif intencao['label'] == 'LABEL_2':
        return "Posso verificar a previsão do tempo para você."
    else:
        return "Desculpe, não entendi. Pode reformular?"
    
resultado = chatbot_bert("Olá!, como vai você?")
print(resultado)
```
